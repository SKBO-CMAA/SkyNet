# Genera los eventos en json a partir de un archivo de excel


```python
import pandas as pd
import json

# Leer la hoja específica del archivo Excel
excel_path = r"E:\SkyNet\eventos.xlsx"

# Leer Excel
df = pd.read_excel(excel_path, sheet_name='Hoja1')
df 
```

```python

# Mapa de contactos por localidad
contactos = {
    "Fontibón": "Sandra Rodriguez 3209163105",
    "Engativá": "Yisel Sierra 3102803999",
    "Funza": "Carolina Villarraga 3209163131"
}



# Normalizar nombres de columnas
df.columns = df.columns.str.strip().str.lower()

resultado = []

for _, fila in df.iterrows():

    localidad = str(fila.get("localidad", "")).strip()

    # Convertir fecha a formato YYYY-MM-DD
    fecha = pd.to_datetime(fila["fecha"]).strftime("%Y-%m-%d")

    registro = {
        "actividad": str(fila.get("actividad", "")).strip(),
        "fecha": fecha,
        "hora": "",
        "localidad": localidad,
        "barrio": str(fila.get("barrio", "")).strip(),
        "direccion": "",
        "contacto": contactos.get(localidad, ""),
        "enlace": ""
    }

    resultado.append(registro)

# Guardar como JSON
with open("actividades.json", "w", encoding="utf-8") as f:
    json.dump(resultado, f, ensure_ascii=False, indent=2)

# Guardar como TXT con el mismo formato
with open("actividades.txt", "w", encoding="utf-8") as f:
    for item in resultado:
        f.write(json.dumps(item, ensure_ascii=False, indent=2))
        f.write(",\n")

print(f"Se generaron {len(resultado)} registros.")
```
